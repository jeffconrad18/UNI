Python also knows the quantifier "all" as `all` and "some" as `any`.

`all` and `any` take a set (more generally, an "iterable") and a predicate and check if all resp. some element has the property specified by the predicate.

We first define a set $S$. In mathematics, we would write $S:=\{0,1,2,3,4,5,6,7,8\}$, in Python this becomes:

```python
S = {0,1,2,3,4,5,6,7,8} # Alternatively: `S = set(range(9))`
print(S)
print(set(range(9)))
```

The quantified statement "all $x$ in $S$ are positive" has to be written as

```python
all(x > 0 for x in S)
```

Python essentially replaces `x` by every element of `S` and checks if `x>0` is true or false. Python does not return a "counterexample" or "witness".

Here are more examples:

```python
S = {0, 1, 2, 3, 4, 5, 6, 7, 8}
print(f"S={S}")
print("All x in S are positive:", all(x > 0 for x in S))
print("Some x in S are positive:", any(x > 0 for x in S))
print("It is not true that all x in S are not positive:", not all(not x > 0 for x in S))
print("It is not true that some x in S are not positive:", not any(not x > 0 for x in S))
print("Some x in S is positive and less than 3:", any(x > 0 and x < 3 for x in S))
print("For all x in S there is some y in S which is equal to x:", all(any(x == y for y in S) for x in S))
print("There is some y in S so that all x in S are equal to y:", any(all(x == y for x in S) for y in S))
```

Note that `all(x == y for x in S)` itself cannot be evaluated:

```python
all(x == y for x in S)
```

While the variable `x` is bound to be an element of `S`, the variable `y` is free. Conceptually `all(x == y for x in S)` is thus a predicate in the variable `y`.

## Sets

In mathematics, we write $\{x \in A \mid P(x)\}$ to select from an already defined set $A$ the elements which have property specified by the predicate $P$, i.e.:

- $y\in \{x\in A \mid P(x)\}$ **:iff** $y\in A$ and $P(y)$ is true.

In Python, this becomes `{x for x in A if P(x)}`.

Similarly, $\{e(x) \mid x\in A\}$ is used to define a new set by iterating over the already defined set $A$ and collecting the results we obtain by evaluating an expression $e$ for every concrete element of $A$, i.e.:

- $y\in \{e(x) \mid x\in A\}$ **:iff** there is some $x\in A$ so that $e(x)$ evaluates to $y$.

In Python, this becomes `{e(x) for x in A}`.

> An expression is something like `x+y` or `2*x` but also $\{x\}$.

Below, we first define the set $A:=\{0,1,2,3,4\}$ by explicitly listing each element.

Afterwards we define the following sets:

- $B :=\{x \in A \mid x \le 2\} = \{0,1,2\}$
- $C :=\{x \in A \mid x \textsf{ even }\} = \{0,2,4\}$
- $D :=\{2x\mid x\in A\}$
- $E :=\{\{x\}\mid x \in A\}$

In mathematics, a set is defined once, in Python we can update a set also afterwards.
We have to tell Python by means of `frozenset` that a set will not change anymore, i.e. that its definition is final, before we can use it as an element within another set.

```python
A = {0, 1, 2, 3, 4}              # $A :=\{0,1,2,3,4\}$
B = {x for x in A if x <= 2}     # $B :=\{x \in A \mid x \le 2\} = \{0,1,2\}$
C = {x for x in A if x % 2 == 0} # $C :=\{x \in A \mid x \textsf{ even }\} = \{0,2,4\}$
D = {2*x for x in A}             # $D :=\{2x\mid x\in A\}$
E = {frozenset({x}) for x in A}  # $E :=\{\{x\}\mid x \in A\}$
print(A,B,C,D,E)
```

The following starts with the empty set $\{\}$ and "grows" it step by step by adding the numbers $0,1,2,3,4$ in the given order.

```python
F = set()                        # $F := \{\}$
for i in range(5): F.add(i)      # for $i$ from $0$ to $4$: $F := F \cup \{i\}$
A == F                           # Is it true that A and S are equal?
```

The following show how the standard set operations and relations can be expressed in Python:
- $A\cup B$: `A|B`, `A.union(B)`
- $A\cap B$: `A&B`, `A.intersection(B)`
- $A\setminus B$, $A-B$: `A-B`, `A.difference(B)`
- $A\Delta B$: `A ^ B`, `A.symmetric_difference(B)`
- $A\subseteq B$: `A <= B`, `A.issubset(B)`
- $A\subsetneq B$: `A < B`
- $A=B$: `A==B`

```python
A = {0,1}; B = {0,2}; C = {0}
print(len(A))                           # $|A|$
print(A | B, A.union(B))                # $A\cup B$
print(A & B, A.intersection(B))         # $A\cap B$
print(A - B, A.difference(B))           # $A\setminus B$
print(A ^ B, A.symmetric_difference(B)) # $A\Delta B$
print(A <= B, A.issubset(B))            # $A\subseteq B$
print(A <= A, A.issubset(A))            # $A\subseteq A$
print(A < A)                            # $A\subsetneq A$
print(C < A)                            # $C\subsetneq A$
print(A == {1,0})                       # Order does not matter
```

In [set theory](https://en.wikipedia.org/wiki/Zermelo%E2%80%93Fraenkel_set_theory#Axiom_of_union_-_5), the union is actually defined for a set of sets, i.e. if every $S$ is set so that every $X\in S$ is again a set, then

- $y\in \bigcup S$ **:iff** there is some $X\in S$ with $y\in X$.

Also the notation $\bigcap S$ is often used

- $y\in\bigcap S$ **:iff** for every $X\in S$ we have $y\in X$.

**but** this is only a well-defined **set** if $S\neq \emptyset$.

> If $S=\emptyset$, then "$for all X\in \emptyset$ we have $y\in X$" is always true as there is no $X\in \emptyset$ for which $y\in X$ is false. I.e. the result would be a collection of "every thing" which is at least not a set.
>
> Using the powerset (see below) we have $\bigcap S=\{x\in\bigcup S\mid \text{for all } X\in S: x\in X\}$.
>
> (Note that the RHS becomes the empty set if $S=\emptyset$.)


If $S=\{X_1,\ldots,X_n\}$ is a finite set with each $X_i$ a set and $n\ge 1$, this amounts to

- $\bigcup S = X_1\cup X_2 \cup \ldots \cup X_n$.
- $\bigcap S = X_1\cap X_2 \cap \ldots \cap X_n$.

We can define this in Python as follows:

```python
def union(S: set[frozenset]):
  return set().union(*S)

def intersection(S: set[frozenset]):
  M1, *S = S
  return {x for x in M1 if all(x in M for M in S)}

A = {0,1}; B = {0,2}; C = {0}
print(union({frozenset(A),frozenset(B), frozenset(C)}))
print(intersection({frozenset(A),frozenset(B), frozenset(C)}))
```

The power set $2^M$ of a set $M$ is the set of all subsets of $M$:

- $x\in 2^M$ **:iff** $X\subseteq M$

Note, that we always have $\emptyset \in 2^{M}$ and $M\in 2^{M}$.
> Alternatively: $\{\emptyset,M\}\subseteq 2^M$.

If $M=\{x_1,\ldots,x_n\}$ is a finite set, we can compute $2^M$ "recursively":
- if $n=0$, i.e. $M=\emptyset$, we have $2^M=2^{\emptyset}=\{\emptyset\}$.
- If $n>0$, we can pick any $x\in M$, and first compute $2^{M-\{x\}}$ and then return $2^{M-\{x\}} \cup \{ K\cup\{x\} \mid K\in 2^{M-\{x\}}\}$.

```python
def powerset(M: set) -> set[frozenset]:
  if len(M) == 0: return {frozenset()}
  x, *M_x = M
  powM_x = powerset(M_x)
  return powM_x | {K | {x} for K in powM_x}

print(powerset(set()))
print(powerset({0}))
print(powerset({0,1}))
```

## Sequences

A *set* $S$ is a *collection* for which we can **only** ask if some "thing" $x$ is an element of $S$ ($x\in S$).
>A set does **neither** know the concept of order (1st, 2nd, 3rd ... element) **nor** that of multiplicity (how often is $x$ an element of $S$?).

An *infinite sequence $s$* is a *collection* where we can ask what its $i$-th entry (component, item, member, element) $s_i$ for every $i\in\N$ (or for every $i\in\N_0$).
- $i$ is the index ("pointer", "address"); most programming languages use $0$ as first index.

A *finite sequence $s$ of length $n$* is a *collection* where we can ask what its $i$-th entry (component, item, member, element) $s_i$ for every $i\in[n]=\{1,2,\ldots,n\}$ (or for every $i\in\Z_n=\{0,1,\ldots,n-1\}$).
- A sequence of length $n$ is also called a *$n$-tuple*.
- A $2$-tuple is also called a *(ordered) pair*.


In mathematics, sequences are defined using round brackets (parentheses), e.g.,
$$(0,1,2,3).$$
Infinte sequences are usually defined by means of expressions $e(i)$ which evaluate to some well-defined "thing" for every natural number $i\in\N$ ($i\in\N_0$), e.g.,
$$(i^{-1})_{i\in\N}=(1,1/2,1/3,1/4,...).$$
- A sequence/tuple of length $0$ is called the *empty tuple* and denoted by $()$.
> A sequence/tuple is an enumeration or listing of objects where objects can be mentioned multiple times, e.g., when we want to represent measurements over discrete points in time where $s_i$ is the measurement at time $t_i$.

Just as sets also sequences and tuples are standard *abstract data types*, i.e. there exist multiple ways to make the computer store sets and sequences with different advantages and disadvantages (depending on the interal "data structure").

Python knows "updatable" (mutable) sequences as `list`, while *unmodifiable* (immutable, fully defined) *finite* tuples are also called `tuple` in Python.

*Strings* ("words") are sequences/tuples of letters (alphabet symbols, characters) where we simply drop the brackes and the commata. *Strings* are called `str` in Python.

A Python `list` is defined using `[...]` while a Python `tuple` is defined using `(...)` (as in mathematics). Strings are defined using either `"..."` or `'...'`.
>  The matching quotation marks tell Python that it should not try to treat the enclosed text as a variable name.

To access the $i$-th entry of a Python sequence, append `[i]`.

Python also allows negative indices `-i` ($i>0$): these are read as "$i$-th entry from the right", so `[-1]` the last/right-most entry, `[-2]` the second to last entry, and so on.

You can also tell Python to that you want a sequence which only consists of the entries
- starting at (including) index `i` up to (excluding) index `j` using `[i:j]`.
- up to (excluding) index `j` using `[:j]`.
- starting at (including) index `i` using `[i:]`.



Some examples:
- `e=()` assigns to `e` the empty tuple.
- `a=tuple(i for i in range(5))` (or simply `tuple(range(5))`) yields $(0,1,2,3,4)$.
- `b=(e,)` tells Python to wrap the empty tuple into a tuple of length $1$ (the trailing `,` is necessary as Python otherwise unwraps it $1$-tuples immediately).
- `w='ABCD'` assigns the string/word "ABC" to `w`.
- `(e,a,b,w)` is itself a $4$-tuple. By means of `for` we can iterate of the tuple from the first to last entry.

```python
e = ()                                # empty sequence/tuple
a = tuple(i for i in range(5))        # $a=(0,1,2,3,4)=(i)_{i\in\Z_5}$
b = (e,)                              # $(())$ (Python needs the trail `,`)
w = 'ABCD'                            # words/strings are sequences/tuples of letters
for s in (e,a,b,w):                   # sequences can be elements of sequences
  print(f"s={s}, |s|={len(s)}")
  for i in range(len(s)):             # entry at i, prefix of length i, suffix starting at i
    print(f"  s[{i}]={s[i]}, s[:{i}]={s[:i]}, s[{i}:]={s[i:]}")
  for i in range(1,len(s)+1):         # a[-1]:=a[len(a)-1], a[-2]:=a[len(a)-2], ...
    print(f"  s[{-i}]={s[-i]}")
```

A `tuple` (or `str`) cannot be updated, i.e., you can read its $i$-th entry, but you cannot change it.


```python
s = "abc"
print(f"s[0]={s[0]}")
s[0] = "x"
print(f"s[0]={s[0]}")
```

Here we fist "grow" the geometric sequence, then update it so that it becomes the geometric sum to avoid recomputation.

```python
geo = [1]
for i in range(10):
  print("Current value of `geo`", geo)
  geo.append(geo[-1] * (-0.5))          # In lists we can also redefine the entry at position/index i
print(f"Geometric sequence: {geo}")
for i in range(1,len(geo)):
  print("Current value of `geo`", geo)
  geo[i] += geo[i-1]
print(f"Geometric sum: {geo}")
```

## Induction

"Induction" can have several related meanings:
- An inductive inference where based on finitely many observations we guess what the underlying "rule" might be.
- A "proof by induction" refers to the definition (axiom of induction) of the natural numbers $\N_0$ and is used to formally argue whether some predicate is true for all natural numbers.
- An "indutive definition" uses the axiom of induction to define, e.g., an infinite sequence.

Below, we recompute the first 12 Fibonacci numbers:
- We explicitly define the first two Fibonacci numbers $F_0:=0$, $F_1:=1$
- Afterwards we use the "defining rule" $F_{n+2}:=F_n+F_{n+1}$ to compute the remaining numbers.

We then compute the sequence of sum of the first $n$ Fibonacci numbers.

```python
fib = [0,1]
for i in range(10):
  fib.append(fib[-1] + fib[-2])
print(f"Fibonacci numbers: {fib}")

sum_fib = [fib[0]]                    # Colab might read `sum_fib` and complete the rest of the code
for i in range(1,len(fib)):
  sum_fib.append(sum_fib[-1] + fib[i])
print(f"Sum of Fibonacci numbers: {sum_fib}")

import matplotlib.pyplot as plt
plt.plot(fib, label="$F_n$")
plt.plot(sum_fib, label="$\\sum_{i=0}^n F_i$")
plt.legend()
plt.show()

# Compare the two sequences.
# How might `fib` and `sum_fib` be related?
# (Drop the first two entries of `fib`.)
# Guess the general rule from these examples (inductive reasoning).
# Then prove your conjecture.
```

The sum of first Fibonacci numbers are our observation.

By comparing the two sequences, we can come up with the "guess" that the sequence of the sum of the Fibonacci numbers is essentially the sequence of Fibonacci numbers, we only need to drop the first two Fibonacci numbers; this "guess" can be "validated" using a proof by induction.

## Relations

A *relation* is a subset $R\subseteq A_1\times\ldots\times A_n$ of some Cartesian product of sets $A_1,\ldots,A_n$.

Relations are the basis of "relational databases" and also the formal treatment "first-order logic".

In principle, it suffices to consider only binary relations $R\subseteq A\times B$ for two sets $A,B$.

A relation $R\subseteq A\times B$ can be visualized by drawing for each element of $A\cup B$ a "dot" into the plane and then drawing for each $(a,b)\in R$ an arrow from the dot representing $a$ to the dot representing $b$.

Below is some simple code to visualize some relations for you.
> All relations defined on $\Omega=\{1,2,3,4\}$, i.e., they are subsets of $\Omega\times\Omega$.
>
> A relation $R\subseteq A\times A$ is also called a "(endo)relation on $A$".
>
> $\textsf{id}:=\{(i,i)\mid i\in \Omega\}$ the *identity* relation on $\Omega$.
>
> These should be the relations also used in the slides as example for the different properties of endorelations.

```python
import networkx as nx
import matplotlib.pyplot as plt
from itertools import product
from typing import *

def plot_rel(R: Set[Tuple[Any,Any]], Universe: Optional[Set] = None, pos=None):
  print("Relation:", R)
  G = nx.DiGraph()
  if Universe is not None:
    G.add_nodes_from(Universe)
  G.add_edges_from(R)
  if pos is None:  # assign positions if necessary
    pos=nx.kamada_kawai_layout(G, pos=nx.spring_layout(G, seed=42))
  nx.draw(G, with_labels=True, pos=pos)
  plt.show()

Omega = {1,2,3,4}
ID = {(1,1),(2,2),(3,3),(4,4)}
R1 = {(1,2),(2,3),(3,1),(4,4)}
R2 = R1 | {(1,3),(3,2),(2,1)}
R3 = {(1,2),(2,3),(3,4)}
R4 = R3 | {(1,3),(1,4),(2,4)}
R5 = R4 | {(1,1),(2,2),(3,3),(4,4)}

pos = {1: (0,0), 2: (0,1), 3: (1,0), 4: (1,1)}  # positions as in slides
for R in (ID,R1,R2,R3,R4,R5):
  plot_rel(R,Omega,pos=pos)
```

Here are the definitions of standard properties (predicates) used to characterize endorelations.

Note that `is_reflexive` and `is_irreflexive` depend both on the relation $R$ and the set $A$ the relation is defined.

All the other properties only depend on the relation as set of pairs.

```python
from itertools import product

is_reflexive = lambda R, Universe: all( (a,a) in R for a in Universe)
is_irreflexive = lambda R, Universe: all( (a,a) not in R for a in Universe)
is_symmetric = lambda R: all( (b,a) in R for (a,b) in R)
is_antisymmetric = lambda R: all( (b,a) not in R for (a,b) in R if a != b)
is_asymmetric = lambda R: all( (b,a) not in R for (a,b) in R)
is_transitive = lambda R: all( (a,c) in R for (a,b1),(b2,c) in product(R, repeat=2) if b1==b2)

# Examples from slides
Omega = {1,2,3,4}
ID = {(1,1),(2,2),(3,3),(4,4)}
R1 = {(1,2),(2,3),(3,1),(4,4)}
R2 = R1 | {(1,3),(3,2),(2,1)}
R3 = {(1,2),(2,3),(3,4)}
R4 = R3 | {(1,3),(1,4),(2,4)}
R5 = R4 | {(1,1),(2,2),(3,3),(4,4)}

for R in (ID,R1,R2,R3,R4,R5):
  print(f"R={R}")  # Colab might (try to) take over again
  print(f"  reflexive    : {is_reflexive(R,Omega)}")
  print(f"  irreflexive  : {is_irreflexive(R,Omega)}")
  print(f"  symmetric    : {is_symmetric(R)}")
  print(f"  antisymmetric: {is_antisymmetric(R)}")
  print(f"  asymmetric   : {is_asymmetric(R)}")
  print(f"  transitive   : {is_transitive(R)}")
  print(f"  eq. relation : {is_reflexive(R,Omega) and is_symmetric(R) and is_transitive(R)}")
  print(f"  partial order: {is_reflexive(R,Omega) and is_antisymmetric(R) and is_transitive(R)}")
```

## Functions

### As set of pairs

A function $f\colon A\to B$ is formally a particular relation $f\subseteq A\times B$ where every $a\in A$ is related to exactly one $b\in B$
- i.e., for every $a\in A$, there is some $b\in B$ with $(a,b)\in f$; and whenever we have both $(a,b)\in f$ and $(a,b')\in f$, then $b=b'$.

This unique $b\in B$ with $(a,b)\in f$ is denoted by $f(a)$ and called the image of $a$ w.r.t. $f$.

> When visualizing $f$ as done above, this means that every $a\in A$ points at exactly one $b\in B$, namely $f(a)$.

In mathematics, we distinguish between the expression used to define a function, and the function itself.

E.g., "$f\colon \R\to\R, x\mapsto x^2$" is short for $\{(x,x^2)\mid x\in \R\}$ as subset of $\R\times \R$.

The expression $x^2$ itself is not a function, as a function always requires domain $A$ and a codomain $B$.

E.g., properties like "injective", "surjective" and "bijective" depends on $A$ and $B$.
- $f_1\colon \R\to\R,x\mapsto x^2$ is neither injective nor surjective.
- $f_2\colon \R_{\ge0}\to\R,x\mapsto x^2$ is injective, but not surjective.
- $f_3\colon \R\to\R_{\ge0},x\mapsto x^2$ is surjective, but not injective.
- $f_4\colon \R_{\ge0}\to\R_{\ge0},x\mapsto x^2$ is bijective.

Functions are visualized in different ways.
1. We can visualize $f\colon A\to B$ as any general relation as above.
2. To also distinguish $A$ and $B$, we can draw to distinct "dot clouds" (one for $A$, typically on the left, and one for $B$, then on the right) and draw for each $(a,b)\in f$ an arrow from the dot for $a$ on the left (before applying $f$) to the dot for $b$ on the right (after applying $f$).
3. If $A,B\subseteq\R$, we can instead treat each $(a,b)\in f$ as the coordinate of a point in the Euclidean plane.

1. and 2. are useful for general (finite) functions, 3. is more useful for functions on $\R$.


```python
import networkx as nx
import matplotlib.pyplot as plt
from typing import *

f_def = lambda x: x + 1      # just $x\mapsto x+1$, not yet a function in the strict sense
g_def = lambda x: (x+1) % 5  # $\text{(see \href{https://en.wikipedia.org/wiki/Lambda_calculus\#Lambda_terms}{lambda calculus/terms})}$

domain = {x for x in range(10)}
# domain = {x/2 for x in range(20)}  # for comparison

f = {(x,f_def(x)) for x in domain}
g = {(x,g_def(x)) for x in domain}

def plot_func(func_def: Callable[[Any], Any], domain: Set[Any], codomain: Set[Any] | None=None):
  f = {(x,func_def(x)) for x in domain}
  # for a finite function, we can easily determine its image:
  image = {y for (x,y) in f}
  # in general, image of f is not computable,
  # hence we sometimes need to use a larger (but simpler) codomain
  # `surjective` is defined w.r.t. the codomain
  if codomain is None:
    print("Using image as codomain.")
    codomain = image
  else:
    assert image <= codomain

  print(f"Domain of f: {domain}")
  print(f"Codomain of f: {codomain}")
  print(f"Image of f: {set(y for (x,y) in f)}")

  print(f"f={f}")
  # f as table/two-line notation
  f = sorted(f)  # arrange set into sequence sorted w.r.t. preimage
  print(f"As table")
  print(f"x={[x for (x,y) in f]}")
  print(f"y={[y for (x,y) in f]}")

  # as bipartite graph, with domain left (t=0) and image right (t=1)
  G = nx.DiGraph()
  pos = dict()
  for i, x in enumerate(sorted(domain)):
    G.add_node(f"${x}_0$")
    pos[f"${x}_0$"] = (0,i)
  for i, y in enumerate(sorted(codomain)):
    G.add_node(f"${y}_1$")
    pos[f"${y}_1$"] = (2,i)
  for (x,y) in f:
    G.add_edge(f"${x}_0$", f"${y}_1$")
  nx.draw(G, pos, with_labels=True)
  plt.show()

  print(f"f is injective: {all(len(G.pred[f"${y}_1$"]) <= 1 for y in codomain)}")
  print(f"f is surjective: {all(len(G.pred[f"${y}_1$"]) >= 1 for y in codomain)}")

  # as directed graph on the union of domain and image
  G = nx.DiGraph()
  G.add_edges_from(f)
  nx.draw(G, with_labels=True, pos=nx.kamada_kawai_layout(G, pos=nx.spring_layout(G, seed=42)))
  plt.show()

  # as points in the plain
  plt.plot([x for (x,y) in f], [y for (x,y) in f], 'x')
  # plt.plot([x for (x,y) in f], [y for (x,y) in f], ':')
  plt.show()

plot_func(f_def, domain)
plot_func(g_def, domain)
```

To visualize a function $f\colon A\to B$ with $A,B\subseteq \R$, we usually implicitly assume that it is continuous, and actually approximate it by draw sufficently many points $(a,f(a))$ into the plane and then connect these points using lines.

You computer essentially does the same.

If $f\colon A\to B$ is bijective, also $f^{-1}=\{(f(a),a)\mid a\in A\}\subseteq B\times A$ is a function from $B\to A$.

```python
# sin, arcsin
import matplotlib.pyplot as plt
import math

#xl = -2*math.pi
xl = -math.pi/2
xr = -xl
n = 100
# Compute 100 values between -pi/2 and pi/2
xs = [xl + (xr-xl)/n * i for i in range(n+1)]
# Compute the corresponding images
ys = [math.sin(x) for x in xs]
# Tell matplotlib to draw the points and connect them using lines segments (green)
plt.plot(xs,ys,color='green')
# The same, only with x/y swapped (red)
plt.plot(ys,xs,color='red')
# The identity function as symmetry axis (black)
plt.plot(xs,xs,'--',color='black')
plt.show()
```

## Real numbers

Eventually, irrational numbers are usually approximated by means of a rational number which is sufficiently close.

While most irrational numbers cannot be approximated using the computer,
for many important constants and functions we have algorithms for their approximation.

Often, the approximation is iteratively improved by adding (eventually vanishing) corrections to the current approximation. This is typically denoted as infinite sum (a series).

Here are the examples from the slides:

- $\sum_{n=0}^{\infty}\tfrac{1}{n!}=e\approx 2.718281828459045$.
- $\sum_{n=0}^{\infty}\tfrac{1}{(n+1)^2}=\tfrac{\pi^2}{6}\approx 1.6449340668482264$
- $\sum_{n=0}^{\infty}\tfrac{1}{4n+1} - \tfrac{1}{4n+3}=\tfrac{\pi}{4}\approx 0.7853981633974483$.
- $\sum_{n=0}^{\infty}\tfrac{1}{2n+1} - \tfrac{1}{2n+2}=\ln(2)\approx 0.6931471805599453$

I.e., you should read series $\sum_{i=0}^\infty a_i$ actually as a small program on how to approximate some value:
- $s_0:=a_0$
- repeat $s_{n+1}:=s_n + a_{n+1}$ until $s_{n+1}$ is sufficiently close to number which we want to approximate.

Central problem: how decide if $s_{n+1}$ is $\varepsilon$-close the unknown $\sum_{n=0}^\infty a_n$? I.e., when to terminate?

```python
import matplotlib.pyplot as plt
import math

n = 40  # change `n`: most of these approximations converge very slowly
approx_e = [0]
approx_pi2 = [0]
approx_pi = [0]
approx_log2 = [0]
for i in range(n):  # Colab tried to complete the code based on the variables names ... and failed.
  approx_e.append(approx_e[-1] + 1/math.factorial(i))
  approx_pi2.append(approx_pi2[-1] + 6*pow(i+1,-2))
  approx_pi.append(approx_pi[-1] + 4 * (pow(4*i+1,-1) - pow(4*i+3,-1)))
  approx_log2.append(approx_log2[-1] + pow(2*i+1,-1)-pow(2*i+2,-1))

print(f"e: {approx_e[-1]} vs {math.exp(1)}")
print(f"pi^2: {approx_pi2[-1]} vs {math.pi ** 2}")
print(f"pi: {approx_pi[-1]} vs {math.pi}")
print(f"ln(2): {approx_log2[-1]} vs {math.log(2)}")

plt.plot(approx_e, label="$e=\\sum_{n=0}^{\\infty} \\frac{1}{n!}$")
plt.plot(approx_pi2, label="$\\pi^2=6\\sum_{n=0}^{\\infty} \\frac{1}{(n+1)^2}$")
plt.plot(approx_pi, label="$\\pi=4\\sum_{n=0}^{\\infty} \\frac{1}{4*n+1}-\\frac{1}{4*n+3}$")
plt.plot(approx_log2, label="$\\ln(2)=\\sum_{n=0}^{\\infty}\\frac{1}{2*n+1}-\\frac{1}{2*n+2}$")
plt.legend()
plt.show()
```

# Limits and series

The following plots a sequence $(a_n)_{n\in\N_0}$ up to some $n_{\max}$ and checks from which index on this "prefix" is $\varepsilon$-close the the conjectured limit $a_{\lim}$.
> As this only considers finitely many values, this cannot be used to decide if $a_{\lim}$ is indeed the limit $(a_n)_{n\in\N_0}$.

```python
from typing import *
from numbers import Number
import matplotlib.pyplot as plt
import math

def plot_limit(a_def: Callable[[int, Number], Number], a_lim: Number, n_max: int, eps: Number):
  a = [a_def(0,0)]
  for n in range(1,n_max):
    a.append(a_def(n,a[-1]))
  print(f"Sequence: a={a}")
  # Determine least n0 so that |a_lim - a_i| < eps for all i >= n_0
  n0 = len(a)
  while n0 > 0 and abs(a[n0-1] - a_lim) < eps:
    n0 -= 1
  print(f"Sequence {eps}-close to {a_lim} starting at index {n0}: {a[n0:]}")
  # Sequence up to index n0-1
  plt.plot(a[:n0],'x',color='red')
  # Sequence for indices >= n0
  plt.plot(list(range(n0,len(a))), a[n0:],'x',color='green')
  # To make the picture nicer
  plt.plot(a,':',color='blue')
  # Upper and lower bounds
  plt.plot([a_lim + eps for _ in range(len(a))],'--',color='orange')
  plt.plot([a_lim - eps for _ in range(len(a))],'--',color='orange')
  plt.show()
  return

# a_def = lambda n, a_prv: 2-1/(n+1)  # a_lim=2
a_def = lambda n, a_prv: 1 if n == 0 else a_prv + (2-a_prv**2)/(4*a_prv) # a_lim=math.sqrt(2)
# x = 0.5
# a_def = lambda n, a_prv: a_prv + pow(x,n)/math.factorial(n)  # a_lim=math.exp(x)
# a_def = lambda n, a_prv: a_prv + (-1)**n * pow(x,2*n+1)/math.factorial(2*n+1)  # a_lim=math.sin(x)
# a_def = lambda n, a_prv: a_prv + (-1)**n * pow(x,2*n)/math.factorial(2*n)  # a_lim=math.cos(x)
plot_limit(a_def, a_lim=math.sqrt(2), n_max=20, eps=0.01)
```

# Representation of numbers

Formally, there is difference between a number $x\in\R$ and the way of how we write/represent it w.r.t. some fixed base $b$; we will assume in the following that $b\in\N$ and $b>1$.

Assume first that $x\in\N_0$. Then we need to determine the digits $d_i\in\{0,1,2,\ldots,b-1\}=\Z_{b}$ s.t.
$$x=d_l \cdot b^l + d_{l-1}\cdot b^{l-1} + \ldots + d_1 \cdot b^1 + d_0$$
Hence, $d_0$ is the remainder of $x$ divided by $b$. Applying this iteratively, we can determine all digits w.r.t. the chosen base $b$:

```python
def represent_natural_in_base(x: int, b: int) -> str:
  assert b > 1 and x >= 0
  if x == 0:
    return "0"
  res = ""
  while True:
    x, r = divmod(x,b)
    res = str(r) + res    # This is most-significant digit first/left-most
    if x == 0:
      break
  return res

print(f"10 in base 2: {represent_natural_in_base(10,2)}")
print(f"10 in base 3: {represent_natural_in_base(10,3)}")
```

If $x\in\Z\setminus\N_0$, we can simply prepend a "minus" `-` to the "sequence/string of digits" $d_ld_{l-1}\ldots d_0$ representing
 $|x|$.

```python
def represent_integer_in_base(x: int, b: int) -> str:
  assert b > 1
  res = represent_natural_in_base(abs(x),b)
  if x < 0:
    res = "-" + res
  return res

print(f"-10 in base 2: {represent_integer_in_base(-10,2)}")
```



It remains to represent the fractional part, i.e., a number $x\in (0,1)$ w.r.t. the chosen base $b$.

Conceptually, we want to determine digits $d_i\in\Z_b=\{0,1,2,\ldots,b-1\}$ s.t.
$$x=d_1 \cdot b^{-1} + d_2 \cdot b^{-2} + d_3 \cdot b^{-3} + \ldots$$
To this end, we can iteratively multiply by $b$ to move the next digit "before the point", i.e.
$$x\cdot b=d_1 + d_2 \cdot b^{-1} + d_3 \cdot b^{-2} + \ldots$$
so that $d_1$ is the integral part of $x\cdot b$.



```python
def represent_fractional_in_base(x: float, b: int, n_digits: int=10) -> str:
  assert b > 1 and 0 < x < 1
  res = "."
  for _ in range(n_digits):
    x *= b
    d, x = divmod(x,1)
    res += str(int(d))
  return res

for n_digits in range(10):
  print(f"0.1 in base 2: {represent_fractional_in_base(0.1,2, n_digits)}")
```

Obviously, with only finitely many digits (as $b$ is assumed to be a natural number), we can only represent rational numbers:
$$d_1 \cdot b^{-1} + \ldots + d_l \cdot b^{-l} = \frac{d_1 \cdot b^{l-1} +\ldots + d_l}{b^l}\in \Q$$
I.e., for every irrational number, we always need to use infinitely many digits. But also rational numbers can require infinitely many digits depending on the chosen base $b$.

But if $x\in (0,1)\cap \Q$, then we will always eventually start repeating the same finite sequence of digits:

```python
def represent_rational_in_base(p: int, q: int, b: int) -> str:
  assert b > 1 and q != 0
  res = ""
  if q < 0:
    p, q = -p, -q
  if p == 0:
    return "0"
  if p < 0:
    res += "-"
    p = -p
  n, p = divmod(p,q)
  res = represent_integer_in_base(n,b)
  res += "."
  memory = []
  while True:
    d, p = divmod(p * b, q)
    if (d,p) in memory:
      break
    memory.append((d,p))
  i = memory.index((d,p))
  res += "".join(str(d) for (d,p) in memory[:i])
  res += "("
  res += "".join(str(d) for (d,p) in memory[i:])
  res += ")"
  return res

print(f"1.1 in base 2: {represent_rational_in_base(11,10,2)}")
```

# Infinite sums and power series

Some functions $f(x)$ can also be represented as "power series" $\sum_{n=0}^\infty a_n x^n$ (at least for certain domains); this means we can approximate $f(x)$ algorithmically by evaluating $\sum_{n=0}^\infty a_n x^n$.

In particular $\exp,\sin,\cos$ can be approximated in this way.

> See the section on Taylor series in the slides, if you want to know how these series can be obtained.


```python
import matplotlib.pyplot as plt
import math

def a(n: int) -> float:
  return 1 / math.factorial(n)

def b(n: int) -> float:
  k, r = divmod(n,2)
  if r == 1:
    return 0
  return (-1)**k / math.factorial(n)

def c(n: int) -> float:
  k, r = divmod(n,2)
  if r == 0:
    return 0
  return (-1)**k / math.factorial(n)

x_min = -math.pi
x_max = math.pi
N     = 100
xs = [x_min + k*(x_max-x_min)/N for k in range(N)]

n_max = 10
for f, d in zip([math.exp, math.cos, math.sin], [a,b,c]):
  plt.plot(xs, [f(x) for x in xs], label=f.__name__)
  for k in range(n_max):
    plt.plot(xs, [sum(d(i) * x ** i for i in range(k)) for x in xs], label=f"max deg={k}")
  plt.legend()
  plt.show()
```

In general, the value to which the series $\sum_{n=0}^{\infty} a_n$ converges to (if at all) depends on the order in which we add up the elements $a_n$.

Formally, $\sum_{n=0}^{\infty} a_n$ has to be read as $s_0:=a_0$, $s_{n+1}:=s_n+a_{n+1}$.

While the order in which we add up finitely many numbers is irrelevant, this is not true anymore if we add up infinitely many numbers:

Below is the alternating harmonic sum $\sum_{n=1}^{\infty} \tfrac{(-1)^{n+1}}{n}$ and two reordered series.

All three "sum up" the sequence $\left(\tfrac{(-1)^{n+1}}{n}\right)_{n\in\N}$ only the time at which some element of the sequence is added to the approximation differs.

So, the "value" (or "meaning") of a series in general depends on the order.



```python
import matplotlib.pyplot as plt
import math
from fractions import Fraction
def a(n: int): # 1/1, -1/2, 1/3, -1/4, 1/5, -1/6, 1/7, -1/8, 1/9
  return Fraction((-1) ** n, (n+1))

# the more odd (even) indices we move to the front, the larger (smaller) the limit
def b(n: int): # 1/1, -1/2, -1/4, 1/3, -1/6, -1/8, 1/5, ...
  k, r = divmod(n,3)
  if r == 0: return Fraction(1,2*k+1)
  if r == 1: return Fraction(-1,4*k+2)
  return Fraction(-1,4*k+4)

def c(n: int): # 1/1, 1/3, -1/2, 1/5, 1/7, -1/4, 1/9, 1/11, -1/8
  k, r = divmod(n,3)
  if r == 0: return Fraction(1,4*k+1)
  if r == 1: return Fraction(1,4*k+3)
  return Fraction(-1,2*k+2)

n_max = 20
for expr in [a,b,c]:
  print([str(expr(n)) for n in range(n_max)])
  partial_sums = [0]
  for n in range(n_max+1):
    partial_sums.append(partial_sums[-1]+float(expr(n)))
  plt.plot(partial_sums, label=f"{expr.__name__}")
for limit in [math.log(2), math.log(2)/2, 3*math.log(2)/2]:
  plt.plot([0,n_max],[limit,limit], ':')
plt.legend()
plt.show()
```

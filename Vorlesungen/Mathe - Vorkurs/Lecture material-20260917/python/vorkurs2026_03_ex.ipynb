# Analytical geometry: Euclidean space

## Basics

Straightforward extension from 2d to 3d; plotting is also already implemented.

```python
from typing import *
import math
import matplotlib.pyplot as plt
import numpy as np

Scalar = float

class Vc3(NamedTuple):
  x: Scalar
  y: Scalar
  z: Scalar

  def __add__(self, other: 'Vc3') -> 'Vc3':
    # return Vc3(self.x+other.x, self.y+other.y, self.z+other.z)
    return Vc3(*[sx+ox for sx, ox in zip(self, other)])

  def __sub__(self, other: 'Vc3') -> 'Vc3':
    # return Vc3(self.x-other.x, self.y-other.y,self.z-other.z)
    return Vc3(*[sx-ox for sx, ox in zip(self, other)])


  def __mul__(self, other: Scalar) -> 'Vc3':
    # return Vc3(self.x*other, self.y*other, self.z*other)
    return Vc3(*[sx * other for sx in self])

  def __truediv__(self, other: Scalar) -> 'Vc3':
    return Vc3(*[sx / other for sx in self])

  def __rmul__(self, other: Scalar) -> 'Vc3':
    return self * other

  def __neg__(self) -> 'Vc3':
    # return Vc3(-self.x, -self.y, -self.z)
    return Vc3(*[-sx for sx in self])

  def __matmul__(self, other: 'Vc3') -> Scalar:
    # return self.x*other.x + self.y*other.y + self.z * other.z
    return sum([sx * ox for sx, ox in zip(self, other)])

  def dot(self, other: 'Vc3') -> Scalar:
    return self @ other

  def snorm(self) -> Scalar:
    return self @ self

  def norm(self) -> Scalar:
    return math.sqrt(self.snorm())

  def cross(self, other: 'Vc3') -> 'Vc3':
    return Vc3(
      self.y * other.z - self.z * other.y,
      self.z * other.x - self.x * other.z,
      self.x * other.y - self.y * other.x
    )

  def __xor__(self, other: 'Vc3') -> 'Vc3':  # '^' as shorthand for the cross product
    return self.cross(other)

  def iszero(self) -> bool:
    return all(math.isclose(sx,0) for sx in self)

  def __str__(self) -> str:
    return str(tuple(self))

  def plot(self, ax, color='red', solid=True, n=100):
    # `solid` and `n` are spurious to make it compatible with the other classes
    ax.scatter(*self, color=color, marker='x')

  pass  # end of class

class Ln3(NamedTuple):  # defined by two points on the line
  a: Vc3
  b: Vc3

  def plot(self, ax, color='blue', solid=True, n=100):
    xyz = [[self.a[i], self.b[i]] for i in range(3)]
    ax.plot(*xyz, color=color, alpha=1)

  pass  # end of class

class Pl3(NamedTuple):  # defined by three points on the plane
  a: Vc3
  b: Vc3
  c: Vc3

  def plot(self, ax, color='green', solid=True, n=100):
    # We create a mesh here so that we can also draw the geometry as a wireframe
    d1 = self.b - self.a
    d2 = self.c - self.a
    u = np.linspace(0, 1, n)
    xyz = [self.a[i] + np.outer(d1[i] * u, np.ones(np.size(u))) + np.outer(np.ones(np.size(u)), d2[i] * u) for i in range(3)]
    if solid:
      ax.plot_surface(*xyz, color=color, alpha=0.5)
    else:
      ax.plot_wireframe(*xyz, color=color, alpha=0.5)

  pass  # end of class

class Sp3(NamedTuple):  # defined by a point as center and a point on the sphere
  c: Vc3
  p: Vc3

  def plot(self, ax, color='orange', solid=True, n=100):
    r = (self.p-self.c).norm()
    u = np.linspace(0, 2 * np.pi, n)
    v = np.linspace(0, np.pi, n)
    # Parametrization (c_x + r * cos(u_i)*sin(v_j), c_y + r * sin(u_i) * sin(v_i), c_z + r * cos(v_j))
    x = self.c.x + r * np.outer(np.cos(u), np.sin(v))
    y = self.c.y + r * np.outer(np.sin(u), np.sin(v))
    z = self.c.z + r * np.outer(np.ones(np.size(u)), np.cos(v))
    if solid:
      ax.plot_surface(x, y, z, color=color, alpha=0.5)
    else:
      ax.plot_wireframe(x, y, z, color=color, alpha=0.5)

  pass  # end of class


def plot_geo_3d(objs: Sequence[Union[Vc3,Ln3,Pl3,Sp3]], solid=True, n: int=100):
  fig = plt.figure()
  ax = fig.add_subplot(projection='3d')
  ax.set_aspect('equal')

  for o in objs:
    o.plot(ax, n=n, solid=solid)
  plt.show()


plot_geo_3d([
    Vc3(-10,-10,-10),
    Sp3(Vc3(1,2,3),Vc3(11,2,3)),
    Ln3(Vc3(-10,-10,-10), Vc3(10,10,10)),
    Pl3(Vc3(-1,-2,-3), Vc3(3,4,0), Vc3(2,-1,4))
    ], solid=True)

plot_geo_3d([
    Vc3(-10,-10,-10),
    Sp3(Vc3(1,2,3),Vc3(11,2,3)),
    Ln3(Vc3(-10,-10,-10), Vc3(10,10,10)),
    Pl3(Vc3(-1,-2,-3), Vc3(3,4,0), Vc3(2,-1,4))
    ], solid=False, n=10)
```

## Exercise: Computing the points of minimal distance on two lines

First orthogonal projection of a point $c$ onto the line $a+\lambda v$.

$p=a+\tfrac{\langle v, c-a\rangle_2}{\lVert v\rVert_2^2} v = a + \tfrac{vv^\top}{\lVert v \rVert_2^2} (c-a)$

> Remark: Distance vector
>
> $c-p=\left(\textsf{Id} - \tfrac{vv^\top}{\lVert v\rVert_2^2}\right)(c-a)$
>
> With $n:= \tfrac{v}{\lVert v\rVert_2}$: $\left(\textsf{Id} - n n^\top \right)(c-a)$

Applying Pythagoras on to the right triangle defined by $a,c,p$:

$\lVert c-p\rVert_2^2=\lVert{c-a}\rVert_2^2 - \lVert{p-a}\rVert_2^2 = {\lVert c-a\rVert_2^2 - \tfrac{\langle v, c-a\rangle^2_2}{\lVert v\rVert_2^2}}$

Second, replace $c$ by $c+\mu w$ to obtain:

$\lVert c+\mu w-p\rVert_2^2={\lVert c-a+\mu w\rVert_2^2 - \tfrac{\langle v, c-a+\mu w\rangle^2_2}{\lVert v\rVert_2^2}}$

Expanding:

$\lVert c-a\rVert_2^2 + 2\mu \langle c-a,w\rangle_2 + \mu^2 \lVert w\rVert_2^2 - \tfrac{\langle v, c-a\rangle^2_2 + 2\mu\langle v, c-a\rangle_2 \langle v, w\rangle_2 + \mu^2\langle v, w\rangle^2_2}{\lVert v\rVert_2^2}$

Derivative w.r.t. $\mu$:

$2\mu \lVert w\rVert_2^2 + 2 \langle w,c-a\rangle_2 - 2\mu \tfrac{\langle v,w\rangle_2^2}{\lVert v\rVert_2^2}-2\tfrac{\langle{v,w}\rangle_2 \langle{v,c-a}\rangle_2}{\lVert v\rVert_2^2} \stackrel{!}= 0$

$\mu =  - \tfrac{\langle w,c-a\rangle_2 \lVert v\rVert_2^2 - \langle{v,w}\rangle_2 \langle{v,c-a}\rangle_2}{\lVert w\rVert_2^2 \lVert v\rVert_2^2 - \langle v,w\rangle_2^2}$

## Exercise: Intersecting a line and a plane

The [cross product](https://en.wikipedia.org/wiki/Cross_product) (also "vector product" as the product is a vector) $a\times b$ of two vectors in $\mathbb{R}^3$ yields a vector which is orthogonal to both (if $a,b$ are not linearly dependent).

Given a plane defined by $a,b,c$, we thus obtain a Hesse normal form $\langle n,x\rangle_2 + \delta=0$ for

+ $n=a\times b$
+ $\delta = -\langle n, a\rangle_2$

> *Remark*
>
> Formally, the cross product is defined by
>
> $a \times b = \begin{pmatrix}
a_y b_z - a_z b_y\\
a_z b_x - a_x b_z\\
a_x b_y - a_y b_x
\end{pmatrix}$
>
> One can check by brute-force expansion and simplification that $\langle a, a\times b\rangle_2 = 0$ but there are more elegant arguments, see [here](https://en.wikipedia.org/wiki/Cross_product#Geometric_meaning):
>
> $a \times b$ is defined in such a way that for every vector $v$ the volume of the parallelepiped defined by $v,a,b$ is equal to $\langle v, a\times b\rangle_2$, i.e. $\lVert a\times b\rVert_2$ is the area of the parallelogram $0,a,b,a+b$ and $a\times b$ is orthogonal to the plane defined by $0,a,b$.


```python
def plane_hnf(pl: Pl3):
  a, b, c = pl
  n = (b-a) ^ (c-a)
  return n, -(n @ a)

plane_hnf(Pl3(Vc3(1,2,3), Vc3(3,4,0), Vc3(2,-1,4)))
```

## Circles in three dimensions

Intersecting a plane with a sphere or two spheres will result in general in a circle.

To draw a circle in three dimensions, we need its center $c$, a point $p$ on its circumference, and a normal $n$.

With $r = p-c$ and $s = \tfrac{n}{\lVert{n}\rVert_2} \times s$ we have $\lVert{r}\rVert_2=\lVert{s}\rVert_2$ so that the circle can be parameterized as $c+ r\cdot \cos(\theta) + s\cdot \sin(\theta)$.

```python
class Cr3(NamedTuple):  # defined by a point as center and a point on the sphere
  c: Vc3
  p: Vc3
  n: Vc3

  def plot(self, ax, color='black', solid=True, n=100):
    r = self.p - self.c
    s = (self.n / self.n.norm()) ^ r
    u = np.linspace(0, 2 * np.pi, n)
    xyz = [self.c[i] + r[i] * np.cos(u) + s[i] * np.sin(u) for i in range(3)]
    ax.plot(*xyz, color=color, alpha=1, linewidth=2)
    ax.plot(*[[self.c[i]-self.n[i],self.c[i]+self.n[i]] for i in range(3)])
    ax.plot(*[[self.c[i],self.p[i]] for i in range(3)])

  pass  # end of class

c1 = Cr3(Vc3(0,0,0), Vc3(0,1,0), Vc3(1,0,0))
c2 = Cr3(Vc3(0,0,0), Vc3(1,0,0), Vc3(0,1,0))
c3 = Cr3(Vc3(0,0,0), Vc3(1,0,0), Vc3(0,0,1))
c4 = Cr3(Vc3(1,1,1), Vc3(0,1,1), Vc3(0,-1,1))

# The perspective can be mislead
plot_geo_3d([c1,c2,c3])
plot_geo_3d([c1])
plot_geo_3d([c2])
plot_geo_3d([c3])
plot_geo_3d([c4])
```

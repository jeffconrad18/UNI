# Introduction: solving quadratic equations in $\mathbb{Q}$

```python
from fractions import Fraction      # tell Python to look up what a `Fraction` is
a = 12; b = -12; c = -45            # let $a=12,b=-12,c=-45$
assert a > 0                        # continue onlf if $a>0$ else complain and stop
assert all(isinstance(x,int) for x in [a,b,c]) # assert that $a,b,c\in\Z$
D = b**2 - 4*a*c                    # let $D=b^2-4ac$
k = 0; s = 0                        # start at $k=0$, $s=0$
while s < D: k += 1; s += 2*k-1     # while $k^2 < D$ set $k:=k+1$, then (!) $s:=s+2k-1=k^2-(k-1)^2$
if s == D:                          # $s=k^2=D$ and $k\ge 0$, thus $k=\sqrt{D}$
  sol = {                           # set of solutions:
    Fraction(-b-k,2*a),             # $\tfrac{-b-k}{2a}$
    Fraction(-b+k,2*a)              # $\tfrac{-b+k}{2a}$
  }                                 # (a set does not know multiplicities)
else:
  sol = {}                          # there is no rational solution
print(sol)                          # internal representation
print({str(x) for x in sol})        # pretty printed
```

## `incN`

The superplan `inc` for humans:

- We split the number into the last and the remaining leading digits.
  - If there are no digits, we treat it as "0".
- If the last digit is "9" (the digit of greatest value), we apply our plan to the remaining leading digits and append a "0" to its result.
- Otherwise we replace the last digit by the digit of the next greater value.

And the `incN` superplan now refined for Python:
 > In the comic it is called `incN_rec`.

```python
# THESE ARE LINE COMMENTS, Python ignores this, but the "AI" might read this
# We will represent numbers as strings/words of letters/symbols/characters 
#   so that we can also work in base 16.
# We thus allow the user to tell us what to use as digits:
#   To this end, the user should give us a sequence/tuple/list
#   of his/her choice of digits in increasing order from least to greatest digit.
# If the user does not define the digits,
#   we default to the normal decimal digits: "0", "1", ..., "9"
# NB: inden(ta)tion tells Python which instructions form a "block (group)".
#   This is quite natural to indicate which parts of a text form a group;
#   but most programming languages use brackets for this, recall "I saw a man with a telescope on a hill."
def incN(number: str, digits=None) -> str:
  # If the user does not define the digits, we use the canonical decimal digits
  digits = digits or tuple("0123456789")  # let Python unpack the string into a sequence of symbols
  # If the input is the empty string/word,
  if number == "":
    # we treat it as "0" and thus directly return "1"
    return digits[1]  # done
  # Split the number into last and the remaining leading digits:
  leading_digits, last_digit = number[:-1], number[-1]
  # If the last digit is the greatest digit (the last in the sequence),
  if last_digit == digits[-1]:
    # we need to increment the remaining leading digits and append the first/least digit
     return incN(leading_digits, digits) + digits[0]  # done
  # Else we only need to replace the last digit by its successor in the digit sequence
  # Ask Python to find the index (value) of the last digit
  idx = digits.index(last_digit)
  # If we cannot find the last digit within the allowed digits,
  if idx < 0:
    # we get angry and crash the program to tell the user to pay more attention next time!
    assert False, f"{last_digit} is not an admissible digit, stupid!"
  return leading_digits + digits[idx+1]
```

```python
# Some tests:
def test_incN():
  # We first define our set of potential digits
  # first the canonical/modern decimal digits
  alphabet = tuple(str(i) for i in range(10))
  print(f"Decimal digits: {alphabet}")
  # We extend it by the lower-case Latin letters:
  alphabet += tuple(chr(ord("a")+i) for i in range(26))
  print(f"Extended digits: {alphabet}")
  for base in [10, 2, 16]:
    print("-----")
    digits = alphabet[:base]
    print(f"Base {base} uses {digits} as digits.")
    # Remember: we treat the empty string "" as "0"
    a = ""
    for i in range(20):
      a = incN(a, digits)
      print(a, end=' ')
    print()

test_incN()
```

```python
# Python builts a dictionary which maps a `name` to its meaning/value, e.g. a number or string
# This dictionary is used to "interpret" the code
# The currently defined names can be accessed via `globals`
print("1.", globals())
try:
  print("2.", name)  # If you run this cell again, `name` will be defined (except you restart Python)
except NameError:
  print("2", "name is not defined")
# You can check whether a name is known to Python by means of:
print("3.", "name" in globals())
# After given `name` the meaning "name" (note the different quotation marks)
name = "name"
print("4.", "name" in globals())
#`print(name)` essentially looks up the meaning of "name" first.
print("5.", name)
# Alternatively, you can look it up yourself (usually never needed)
print("6.", globals()["name"])
# We can give `name` a new meaning
name = "awesome new name"
print("7.", name)
# We can use the current definition of `name` also:
name = "super " + name
print("8.", name)
# This works as Python first looks up the meaning/vaue of `name`
# In order to forget a variable, you usually need to restart Python resp. tell Colab to restart from scratch
# A `name` can also refer to a function/procedure
name = lambda: "this returns a string instead of a name"
# The value of `name` is now a function
print("9.", globals()["name"])
# By means of `name()`, Python first looks up the definition, then executes the instructions refered to by `name`
print("10.", name())

# Defining a `name` within a function definition "shadows" the original/outer/global definition:
def some_function():
  name = "local name"  # this is now a "local" `name`; this does NOT change the outer/gobal definition of `name`
  # These use the local definitions:
  # Python keeps nested dictionaries to distinguish local from outer/global definitions
  # Conceptually, Python uses `some_function.name` for the local variable
  # The local meaning of `name`:
  print("11.", name)
  print("12.", locals()["name"])
  # The global meaning of `name`
  print("13.", globals()["name"])
  # Tell Python to execute the function known as `name`
  print("14.", globals()["name"]())
some_function()
```

## Primes and sieve of Eratosthenes

```python
def isprime(n: int) -> bool:
  return n > 1 and all(n % d != 0 for d in range(2,n))

print([n for n in range(100) if isprime(n)])
```

```python
import matplotlib.pyplot as plt
import math
from itertools import accumulate

def plot_pi(n: int):
  # Sieve of Eratosthenes
  is_prime = [True for _ in range(n+1)]
  is_prime[0] = is_prime[1] = False
  for k in range(2,n+1):
    if is_prime[k] is True:
      for j in range(k*k, n+1, k):  # all multiples of $k$ between $k^2$ and $n$
        is_prime[j] = False

  # pi[i]: number of primes in {0,...,i}
  pi = list(accumulate(is_prime))

  # absolute w.r.t. n/log(n)  (natural logarithm)
  plt.plot(list(range(n+1)),pi)
  plt.plot(list(range(n+1)),[i/max(1,math.log(max(i,1))) for i in range(n+1)])
  plt.show()

  # relative w.r.t. n/log(n)  (natural logarithm)
  plt.plot(list(range(1,n+1)),[pi[i]/(i/max(1,math.log(max(i,1)))) for i in range(1,n+1)])
  plt.show()
  return

plot_pi(1000)
```
